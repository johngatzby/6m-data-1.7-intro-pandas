### Challenge 3: Transform, Sort & Rank

**Tasks:**
1. Use `.apply()` to add a `price_tier` column: `'Budget'` (< £35), `'Standard'` (£35–£50), `'Premium'` (> £50).
2. Sort the DataFrame by `revenue` descending.
3. Add a `sales_rank` column using `.rank(ascending=False)`.
4. Use `.map()` to encode `genre` as a number: `{'Tech': 0, 'Data': 1}`.


```python
# MY ANSWER - Challenge 3 #

#1-Use `.apply()` to add a `price_tier` column: `'Budget'` (< £35), `'Standard'` (£35–£50), `'Premium'` (> £50)
def price_tier(price):
    if price < 35:
        return 'Budget'
    elif 35 <= price <= 50:
        return 'Standard'
    else:
        return 'Premium'        

books['price_tier'] = books['price'].apply(price_tier)
display(books)

#2-Sort the DataFrame by `revenue` descending.

books = books.sort_values(by='revenue', ascending=False)
display(books)

#3-Add a `sales_rank` column using `.rank(ascending=False)`.

books['sales_rank'] = books['units_sold'].rank(ascending=False).astype(int)
display(books)  

#4-Use `.map()` to encode `genre` as a number: `{'Tech': 0, 'Data': 1}`
genre_mapping = {'Tech': 0, 'Data': 1}
books['genre_encoded'] = books['genre'].map(genre_mapping)
display(books)
```



<details>
<summary>✅ Solution</summary>

```python
# 1. Price tier
def get_tier(price):
    if price < 35:
        return 'Budget'
    elif price <= 50:
        return 'Standard'
    else:
        return 'Premium'

books['price_tier'] = books['price'].apply(get_tier)

# 2. Sort by revenue
books_sorted = books.sort_values('revenue', ascending=False)

# 3. Sales rank
books['sales_rank'] = books['units_sold'].rank(ascending=False).astype(int)

# 4. Encode genre
books['genre_encoded'] = books['genre'].map({'Tech': 0, 'Data': 1})

print(books[['title', 'price_tier', 'sales_rank', 'genre_encoded']])
```

### 🏆 Stretch Challenge (Optional)

The manager wants a report comparing average price and total revenue by `price_tier`. Use `groupby()` to generate this summary. Which tier contributes the most total revenue despite having the lowest average price?

```python
### 🏆 Stretch Challenge (Optional)

#The manager wants a report comparing average price and total revenue by `price_tier`. Use `groupby()` to generate this summary. 
# Which tier contributes the most total revenue despite having the lowest average price?


summary = books.groupby('price_tier').agg({'price': 'mean', 'revenue': 'sum', 'units_sold': 'sum'})
display(summary)
print("Tier with highest total revenue:", books.groupby('price_tier')['revenue'].sum().idxmax())   
print("Tier with lowest average price:", books.groupby('price_tier')['price'].mean().idxmin())
print("\nNo tier meets both conditions. Budget has the lowest average price, but Standard contributes the most total revenue.")

# Defining a new function to determine the performance of each tier based on ranking of revenue / average price ratio
print("\nBest tier based on revenue to average price ratio:")
def best_tier_new(df, group_col='price_tier'):
    summary = df.groupby(group_col).agg(
        avg_price=('price', 'mean'),
        total_revenue=('revenue', 'sum')
    )
    summary['score'] = summary['total_revenue'] / summary['avg_price']
    summary = summary.sort_values('score', ascending=False)

    winner = summary['score'].idxmax()
    display(summary.round(2))
    print(f"Optimal tier: {winner} "
          f"({summary.loc[winner, 'total_revenue']:,.0f} revenue at "
          f"{summary.loc[winner, 'avg_price']:.2f} avg price)")

best_tier_new(books)
```
